```python
#@title ▶ Download model

# Use a pipeline as a high-level helper
from transformers import pipeline

# llm = pipeline("text-generation", model="HuggingFaceTB/SmolLM2-360M-Instruct")
llm = pipeline("text-generation", model="HuggingFaceTB/SmolLM2-1.7B-Instruct")
```

```python
#@title ▶ Load a dataset (from [Huggingface](https://huggingface.co/datasets))

from datasets import load_dataset

dataset = load_dataset("jeffreyszhou/ikea-us-products-2025")
```

#Processing

```python
from torch import randint
#@title ▶ Process the texts in the dataset with the defined prompt

#@markdown Prompt
prompt = "Given the following product description, describe a house where it fits nicely. Don't name the product, just describe the house." # @param {type:"string"}

#@markdown Amount of texts ro process
amount = 5 # @param {type:"number"}

import random

class color:
   BOLD = '\033[1m'
   END = '\033[0m'


for i in range(min(amount, len(dataset['train']))):
  idx = random.randint(0, len(dataset['train']))

  text = dataset['train']['description'][idx]

  print(f"{color.BOLD}Product name{color.END}: {dataset['train']['title'][idx]}")
  print(f"{color.BOLD}Description{color.END}: {text}")

  messages = [
      {"role": "user", "content": f"{prompt}\n{text}"},
  ]
  output = llm(messages)

  print(f"{color.BOLD}Output{color.END}: {output[0]['generated_text'][1]['content']}")
  print("\n********\n")

```
