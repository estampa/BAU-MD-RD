```python
#@title ▶ Load the model

from ultralytics import YOLO

#@markdown Model size (s, m, x, l) - bigger size means better results, but slower
model_size = 'm' # @param ["s", "m", "l", "x"]


# Initialize a YOLO-World model
model = YOLO(f"yolov8{model_size}-worldv2.pt")  # or select yolov8m/l-world.pt

```

```python
#@title Uncomment to connect drive

# from google.colab import drive
# drive.mount('/content/drive')
```

```python
#@title ▶  Process

#@markdown The output will be saved to "runs/detect/predict"

#@markdown Input the path to a video file, a folder with images or a youtube video link
input_path = "https://www.youtube.com/watch?v=ORrrKXGx2SE" # @param {type:"string"}

#@markdown Enter the classes to be detected, separated by comma
classes = "artist, designer" # @param {type:"string"}

#@markdown Confidence threshold

#@markdown ⚠️ A lower value (like 0.005) will show false detections
score_threshold = 0.005 # @param {type:"number", min:0, max:1}
nms_threshold = 0.7


# Define custom classes
model.to('cpu')
model.set_classes([cl.strip() for cl in classes.split(',')])
model.to('cuda')
print([cl.strip() for cl in classes.split(',')])

results = model.predict(input_path, conf=score_threshold, iou=nms_threshold, save=True, stream_buffer=True)
```
